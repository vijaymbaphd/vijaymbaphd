# 2. Data types


## Numbers: int and float

`int` is a whole number, `float` carries decimals.

Division always returns a float, even when it divides evenly.

```python
x = 10
y = 2.5
print(type(x), type(y))
print(10 / 2)        # 5.0, not 5
print(int(7.9))      # cuts the decimal, does not round
print(float(7))
```

## Complex numbers

Python has a third numeric type, written with `j` for the imaginary part.

Rare in HR work, but it is why `type()` can answer `complex`.

```python
z = complex(0, 5)
print(z, type(z))
print(z.real, z.imag)
```

## Boolean

Two values only, no quotes.

`if` picks a branch. The indentation is the grammar, not decoration.

```python

# Boolean example

# Assigning boolean values
light_on = True
is_raining = False

# Using the boolean values in conditional statements
if light_on:
    print("The light is on.")
else:
    print("The light is off.")

if is_raining:
    print("It's raining. Take an umbrella.")
else:
    print("It's not raining. No need for an umbrella.")
```

## Text

`type()` reports what something is. `str` means text.

```python

my_string = "Hello, Python!"
print(type(my_string))
```

## Upper and lower

String methods return a new string. The original is unchanged.

```python
print("murugan".upper())
print("HI GUYS".lower())
print("hr analytics".title())
```

## Lists

Square brackets. Mixed types allowed.

Counting starts at 0, so `my_list[2]` is the third item.

```python

my_list = [1, 2.5, "word"]
my_list[2]
```

Lists change in place. Assign to a position and the list updates.

```python
my_list[2] = 20
print("Modified List:", my_list)
```

## Tuples

Round brackets. Same as a list, but locked after creation.

```python

my_tuple = (1, 2.5, "word")
print(type(my_tuple))
```

## Sets

Curly braces, no duplicates, no order.

Useful when you only care about which distinct values appear.

```python
depts = {"HR", "Sales", "HR", "IT"}
print(depts)          # HR appears once
print(len(depts))

print(set([1, 2, 2, 3, 3, 3]))   # duplicates collapse
```

## Dictionaries

Key and value pairs. Look up by name instead of position.

`.keys()` and `.values()` show each side.

```python

my_dict = {'name': 'Alice', 'age': 25}
print(type(my_dict))
# Using the .keys() method to view keys
keys = my_dict.keys()
print("keys in the dictionary:", keys)

# Using the .values() method to view keys
values = my_dict.values()
print("values in the dictionary:", values)
```

# 3. Operators

```python
print(5 == 3, 5 != 3)
print(5 > 3, 5 <= 3)
```

# 6. Data manipulation


## Boolean filtering

The condition produces True or False for every row, and the frame keeps the True ones.

Combine conditions with `&` for and, `|` for or. Each condition needs its own brackets.

```python
committed = HR[HR['COM'] > 6.5]
print(committed.shape)

print(HR[(HR['JS'] > 8) & (HR['edu'] == 3)].shape)
```

## np.where

Reads as: condition, value if true, value if false.

This is how a continuous variable becomes a group.

```python
%pip install numpy -q

import numpy as np

HR['high_COM'] = np.where(HR['COM'] > HR['COM'].mean(), 1, 0)
print(HR['high_COM'].value_counts())
```

## Sorting

`ascending=False` puts the largest first.

With two columns, the second breaks ties inside the first.

```python
print(HR.sort_values('COM', ascending=False).head(3))

print(HR.sort_values(['edu', 'COM'], ascending=[True, False]).head(3))
```

## Adding and transforming columns

Arithmetic on a column applies to every row at once. No loop needed.

The z score below puts a variable on a mean of 0 and a standard deviation of 1.

```python
HR['JA_VAL'] = HR['JA'] * HR['VAL']
HR['JA_z'] = (HR['JA'] - HR['JA'].mean()) / HR['JA'].std()

print(HR[['JA', 'VAL', 'JA_VAL', 'JA_z']].head(3))
```

# 7. Aggregation and merging


## groupby and multiple aggregations

Split the rows by a column, then compute something for each group.

`agg` takes a list, so you get several statistics in one table.

```python
print(HR.groupby('edu')['COM'].mean())

print(HR.groupby('sex')[['JA', 'JS', 'COM']].agg(['mean', 'std', 'count']))
```

## Grouping by several columns

Pass a list of columns and pandas splits by every combination.

Watch how thin the groups get. With three demographics the sample is spread across dozens of cells, and a mean built on four people is not worth reporting.

```python
agg = HR.groupby(['sex', 'age', 'edu'])['JS'].mean()
print(agg.head(10))

print(HR.groupby(['sex', 'edu'])['JS'].agg(['mean', 'count']).head(10))
```

## Pivot table and crosstab

`pivot_table` summarises a number across two dimensions.

`crosstab` counts how many people fall in each combination.

```python
print(HR.pivot_table(values='COM', index='edu', columns='sex', aggfunc='mean'))

print(pd.crosstab(HR['edu'], HR['sex']))
```

## Merging

Two frames join on a shared key. Here the key is an employee ID we create.

`_x` and `_y` appear when both sides carry a column of the same name.

```python
left  = HR[['JA', 'COM']].reset_index().rename(columns={'index': 'EID'})
right = HR[['JS', 'VAL']].reset_index().rename(columns={'index': 'EID'})

merged = pd.merge(left, right, on='EID')
print(merged.shape)
print(merged.head(3))
```

# 8. Correlation and regression


## Correlation

`corr()` gives every pair at once, on a scale from -1 to +1.

It measures straight line association only, and says nothing about cause. `pearsonr` adds the p value.

```python
print(HR[['JA', 'VAL', 'JS', 'COM']].corr().round(2))

r, p = stats.pearsonr(HR['JS'], HR['COM'])
print(r, p)
```

## Regression

`ols` reads as outcome ~ predictors.

Three things to read in the output: R-squared, the coefficients, and `P>|t|`. A coefficient is the change in COM for a one point rise in that predictor, with the others held constant.

```python
%pip install statsmodels -q
from statsmodels.formula.api import ols

model1 = ols('COM ~ JA + VAL + JS', data=HR).fit()
print(model1.summary())
```

## Comparing two models

Model A holds only the demographics. Model B adds the job variables.

The jump in R-squared is what the job variables explain beyond who the person is. Both models must be fit on the same rows for the comparison to mean anything.

```python
HR['male'] = np.where(HR['sex'] == 1, 1, 0)

model_a = ols('COM ~ age + edu + male', data=HR).fit()
model_b = ols('COM ~ age + edu + male + JA + VAL + JS', data=HR).fit()

print(model_a.rsquared.round(3), model_b.rsquared.round(3))
print('change in R2:', round(model_b.rsquared - model_a.rsquared, 3))

from statsmodels.stats.anova import anova_lm
print(anova_lm(model_a, model_b))
```
